## Data contract defines your requirements

* A data contract is an agreement between data producers and consumers.
* The producer and consumer can be manual or automated.
* The producer and consumers always has a single responsible team for each.
* Data contract is a document that is human readable. It can be a Google Doc, sheet, yaml, json, Markdown, etc
* A data contract is important as it clearly states what is expected from a dataset. It can be used by various teams to set expectations.

#### Exercise [5 min]

* Assume a stakeholder asks you for a data that is refreshed every 6h. How will you schedule your pipeline?
* Assume the pipeline runtime is 1hour

**Solution**

* When a stakeholder says 'every 6h,' it's critical to clarify what that means.
* While 'every 6h' may suggest that the pipeline should run at least once every 6 hours, it may actually mean that the time between data generation and its appearance in the output data should be no more than 6 hours.
* We schedule the pipeline to run every 3h to give us extra time to debug in case of any issues.
* Another key consideration is the data schema. In particular, remember that your stakeholder may be using your data with a select *. If you change the data schema, it will break their pipelines.

#### Example

* Since the data contract is critical, ensure it is version-controlled and human-readable.
* Here are some key areas to cover when creating a data contract
  1. Data description, & version & how to access the data
  2. Data schema definition,  & grain
  3. Data Quality checks
  4. Data SLA (at least latency)
  5. Data Access
  6. Support for data
