## Outcomes

1. Be able to design reliable data systems. In real projects and in interviews.
2. Understand common problems in data pipelines and startegies to handle them (e.g. Lambda architecture for late events)
3. Showcase expertise and build trust with a potential employer with thoughtfully designed capstone project.  

## A Workshop-style course

1. We achieve this by covering:
      1. Building a Data Warehouse
      2. Designing Data Pipelines with PySpark
      3. Modeling Data with Medallion Architecture
      4. Ensuring Data Quality using soda-core
      5. Scheduling and Orchestrating with Apache Airflow
      6. Testing Code with Pytest
      7. Creating Data Contracts
      8. Completing an Outcome-Focused Capstone Project with Spark, Airflow, and Iceberg using Medallion Architecture
2. The course follows a workshop style, allowing you to run code alongside each chapter.
3. The course consists of **8 sections, 40 chapters, 40+ examples, and 40+ exercises**.
4. The course has over **5 hrs of total video content**.

## Prerequisites

1. Python, SQL basics; Learn these with my free DE101 course links
    * [SQL basics](https://de101.startdataengineering.com/sql_basics)
    * [Window functions](https://de101.startdataengineering.com/windows)
    * [Python basics](https://de101.startdataengineering.com/py_basics)
    * [Pyspark basics](https://de101.startdataengineering.com/py_transform#data-processing-with-pyspark)
2. A machine with at least 4GB of RAM
3. **Windows users**: please setup WSL and a local Ubuntu Virtual machine following [the instructions here](https://documentation.ubuntu.com/wsl/stable/howto/install-ubuntu-wsl2/#what-you-will-learn) until you get an ubuntu prompt.
4. Install [Docker](https://docs.docker.com/engine/install/) and [Docker Compose](https://docs.docker.com/compose/install/)

## Following along

1. Download and unzip the attached code folder.
2. In the extracted folder, run the command `docker compose up -d --build` on your terminal.
3. Wait a few seconds and open JupyterLab at http://localhost:8888.
4. In Jupyter Lab, open the `notebooks` folder, which has one notebook per section. The notebooks with solutions are in the `notebooks_with_solutions` folder.
5. Execute the JupyterLab cells to follow along.
6. Open Airflow at http://localhost:8080, Airflow pipelines are defined at `airflow/dags`
